````python
from IPython.display import Markdown, display
from model_text import display_messages
from llama_index.core.llms import ChatMessage
from model_text import llm_json
from utils import deindent
import json
from pydantic import ValidationError

# Retry count for both steps
MAX_RETRIES = 5

def generate_scenes(story: Story) -> Story:
    """Generates detailed scenes for each act of the story, one act at a time."""

    # Exclude fields not required for act generation
    request_excludes = {
        "characters": {
            "__all__": {
                "animation_description": True,
                "voice_description": True,
                "internal_conflict": True,
                "character_arc": True,
                "voice_sample": True,
                "catch_phrase": True,
                "props": True,
                "image_prompt": True,
                "image_prompt_short": True,
            }
        },
        "story_beats": True,
        "subplots": True,
        "emotional_arc": True,
        "conflict_type": True,
        "motifs": True,
        "narrative_perspective": True,
        "themes": True,
        "props": True,
    }

    for act_idx, act in enumerate(story.acts, start=1):
        display(Markdown(f"### Generating Scenes for Act {act_idx}: {act.title}"))
        total_scenes_so_far = sum(len(a.scenes) for a in story.acts[:act_idx])

        system_message = deindent(f"""
            You are a scene planner. Based on the provided act, your task is to generate a structured list of scenes/chapters for the act.
                                  
            The act currently has {len(act.scenes)} scenes. Generate new scenes to ensure the narrative is cohesive and integrates the story beats, character arcs, and subplots. 
            The total number of scenes across all acts should approximately match the target of {story.avg_scene_count} for the medium: {story.medium}.
            There are {total_scenes_so_far} scenes generated so far in the story.

            Each scene should include:
                - title
                - description
                - time_of_day
                - location
                - lighting
                - mood
                - setting
                - key actions
                - a list of characters involved

            Avoid camera movements or multiple locations in one scene. Return only a valid JSON object with the updated scenes for this act.
        """)

        # Integrate the story beats, character arcs, subplots, and motifs to maintain narrative coherence. Ensure the scenes reflect the themes and emotional arcs of the story.


        # Prepare chat messages
        messages = [ChatMessage(role="system", content=system_message)]
        messages.append(ChatMessage(role="user", content=act.model_dump_json(exclude=request_excludes)))
        display_messages(messages)

        attempt = 0
        while attempt < MAX_RETRIES:
            response = llm_json.stream_chat(messages, max_new_tokens=100000)
            response_text = ""
            for r in response:
                print(r.delta, end="")
                response_text += r.delta

            try:
                # Validate and update the act with new scenes
                updated_act = Act.model_validate_json(response_text)
                act.scenes = updated_act.scenes
                display(Markdown(f":white_check_mark: Act {act_idx} completed with {len(act.scenes)} scenes."))
                break
            except (json.JSONDecodeError, ValidationError) as e:
                attempt += 1
                error_message = (
                    f"**Attempt {attempt} Failed for Act {act_idx}:**\n"
                    f"Error parsing or validating JSON response:\n\n```\n{str(e)}\n```"
                )
                display(Markdown(error_message))
                messages.append(ChatMessage(role="system", content=error_message))
                messages.append(ChatMessage(role="user", content="Here was the invalid JSON attempt:\n" + response_text))

            if attempt >= MAX_RETRIES:
                raise ValueError(f"Failed to generate scenes for Act {act_idx} after {MAX_RETRIES} attempts.")

    display(Markdown(":white_check_mark: Completed scene generation for all acts."))
    return story
````

````python
# Image Prompts

# TODO: Might need to do one scene at a time if the context grows too large and only include the characters involved in the scene

from IPython.display import Markdown, display
from model_text import display_messages
from llama_index.core.llms import ChatMessage
from model_text import llm
from utils import blank_story

request_excludes = {
    "plot_overview": True,
    "characters": {
        "__all__": {
            "character_arc": True,
            "voice_sample": True,
            "catch_phrase": True,
            "animation_description": True,
            "voice_description": True,
            "props": True,
            "relationships": True,
            "internal_conflict": True,
            # "image_prompt": True,
            "image_prompt_short": True,
            "character_arc": True,
        }
    }, 
    "acts": {
        "__all__": {
            "scenes": {
                "__all__": {
                    "scene_image_prompt": True,
                    "scene_image_prompt_short": True,
                    "background_animation": True,
                    "background_image_prompt": True,
                    "dialogue": True,
                }
            }
        }
    },

    "story_beats": True, 
    "subplots":True, 
    "emotional_arc": True, 
    "conflict_type": True, 
    "motifs": True,
    "narrative_perspective": True, 
    "themes": True,
    "props": True,
}


response_includes = {
    "acts": {
        "__all__": {
            "scenes": {
                "__all__": {
                    "title": True,
                    "scene_image_prompt": True,
                    "scene_image_prompt_short": True,
                    "background_animation": True,
                    "background_image_prompt": True,
                }
            }
        }
    },
}

def generate_scene_image_prompts(story: Story) -> Story:
    # TODO: foreground animation
    system_message = (
        "You are an animation director. "
        "Your task is to take a list of acts and scenes and generate the following attributes: background_animation, background_image_prompt, scene_image_prompt, scene_image_prompt_short .\n"
        "\n - background_animation should describe **VISUAL** environmental movement such as fog, dust, bushes rustling, lights flickering, leaves blowing or anything else happening in the background. "
        "\n - background_image_prompt should be a description of the visual elements in the scene with no characters. "
        "\n - scene_image_prompt should be a description of the visual elements in the scene with characters in action as it pertains to the scene. "
        # "\n - scene_image_prompt_short should be a shorter version of `scene_image_prompt` that will fit into 50 words. "
        "\n\n**Do not include camera movements.** \n"
        "Every scene_image_prompt must start with the visual style of the image, for example: 'A digital painting style with watercolor elements of ...'. \n"
        "Make sure the scene_image_prompt starts with the medium, such as painting, cartoon, 8bit, hyperrealistic photo, etc. \n"
        "The visual style of the image must match the visual_style of the story and be based on the story visual_style, time_period, location, genre and medium. \n"
        "Instead of referencing a character name in a scene, describe their physical appearance based on the character image_prompt and physical_appearance so an artist can draw it and animate it without any context. "
        "`scene_image_prompt` can contain layers. \n"
        "**Do not mention character names. Just use their physical descriptions so the illustrator knows what to draw.**\n"
        "For scene_image_prompt describe each character physical appearance in detail in action. \n"
        "Return a json object with the acts and scenes with the title and the newly created attributes background_animation, background_image_prompt, scene_image_prompt, scene_image_prompt_short. \n"
        "Return the same number of acts and scenes as the input. "
        "Refer to `key_actions` to determine and describe the action of the characters in the scene. \n"
        "Return only a valid JSON object. "
    )

    messages = [ChatMessage(role="system", content=system_message)]
    # few_shot_examples_and_blank = [blank_story] + few_shot_examples

    for example in few_shot_examples:
        messages.append(ChatMessage(role="user", content=example.model_dump_json(exclude=request_excludes)))
        messages.append(ChatMessage(role="assistant", content=example.model_dump_json(include=response_includes)))

    # Add the user input (the story prompt)
    messages.append(ChatMessage(role="user", content=story.model_dump_json(exclude=request_excludes)))

    from model_text import display_messages
    display_messages(messages)

    # Call the LLM to generate the complete story with scenes
    response = llm_json.chat(messages, max_new_tokens=100000).message.content

    # resp = llm.stream_chat(messages, max_new_tokens=10000)

    # # Print it out as it comes in
    # for r in resp:
    #     print(r.delta, end="")

    # response = r.message.content

    try:
        story_with_scenes = Story.model_validate_json(response)
        # print("story_with_scenes: ", story_with_scenes.model_dump_json(indent=4))
    except:
        print("Exception occurred while parsing the response.")
        print(response)
        raise
    
    return story_with_scenes
````
