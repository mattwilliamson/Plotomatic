## Generate list of acts

We will feed the LLM a bunch of example inputs and outputs to help guide it.

````python
from IPython.display import Markdown, display
from model_text import display_messages
from llama_index.core.llms import ChatMessage
from model_text import llm, stream_llm_response
from utils import deindent
import json
from pydantic import ValidationError

# Retry count for generating acts
MAX_RETRIES = 5

def generate_acts(story: Story) -> Story:
    """
    Generates the act structure for the story with a focus on high-level narrative arcs.
    Each act should define its purpose, stakes, and how it advances the story.
    """

    # Fields to exclude for act generation
    request_excludes = {
        "characters": {
            "__all__": {
                "animation_description": True,
                "voice_description": True,
                # "internal_conflict": True,
                # "character_arc": True,
                "voice_sample": True,
                "catch_phrase": True,
                "props": True,
                "image_prompt": True,
                "image_prompt_short": True,
            }
        },
        "story_beats": True,
        "subplots": True,
        "emotional_arc": True,
        "conflict_type": True,
        "motifs": True,
        "narrative_perspective": True,
        "themes": True,
        "props": True,
        "acts": True,
    }

    # Fields to exclude from the response
    response_excludes = {
        "characters": True,
        "prompt": True,
        "title": True,
        "visual_style": True,
        "time_period": True,
        "location": True,
        "genre": True,
        "medium": True,
        "plot_overview": True,
        "props": True,
        "acts": {
            "__all__": {
                "chapters": True,
                "props": True,
            }
        },
    }

    system_message = deindent("""
        You are a masterful story architect and narrative designer. Your goal is to craft a compelling act structure for a story based on its core themes, genre, and emotional arc.
        
        The act structure should:
        - Align with the story's genre and narrative style.
        - Highlight the stakes and emotional growth for the protagonist in each act.
        - Emphasize tension, pacing, and resolution that captivate the audience.
        
        Strictly output a structured list of acts. Each act must include and nothing more:
        - **act_id**: A unique identifier for the act.
        - **title**: A creative, thematic title.
        - **description**: The act's purpose and contribution to the narrative.

        Ensure the narrative arc is dynamic and engaging, building to a satisfying climax and resolution.
                              
        

        Provide only a valid JSON object containing the structured acts. Avoid commentary or explanations.
    """)

        # - **Act Title**: A creative, thematic title.
        # - **Purpose**: What this act contributes to the narrative.
        # - **Conflict**: Key challenges and stakes faced during this act.
        # - **Transformation**: How characters or the plot advance by the end of the act.

    # Construct messages for the LLM
    messages = [ChatMessage(role="system", content=system_message)]

    # Optionally add examples for context if few-shot examples are available
    for example in few_shot_examples:
        messages.append(ChatMessage(role="user", content=example.model_dump_json(exclude=request_excludes)))
        messages.append(ChatMessage(role="assistant", content=example.model_dump_json(include=["acts"], exclude=response_excludes)))

    # Add the user story prompt
    messages.append(ChatMessage(role="user", content=story.model_dump_json(exclude=request_excludes)))
    display_messages(messages)

    attempt = 0
    while attempt < MAX_RETRIES:
        response_content = ""

        try:
            # Stream the chat response for dynamic feedback
            response_stream = llm.stream_chat(messages, max_new_tokens=100000)
            response_content = stream_llm_response(response_stream)

            # Parse and validate the response JSON
            story_with_acts = Story.model_validate_json(response_content)
            return story_with_acts

        except (json.JSONDecodeError, ValidationError) as e:
            attempt += 1
            error_message = (
                f"Attempt {attempt} failed: Error parsing or validating JSON response. Error: {str(e)}\n"
                "Please respond with only a valid JSON object containing the structured acts. Avoid extra commentary or explanations."
            )
            print(error_message)

            # Log the failed attempt for debugging
            messages.append(ChatMessage(role="system", content=error_message))
            messages.append(ChatMessage(role="user", content="Here was the invalid JSON attempt:\n" + response_content))

    raise ValueError("Failed to generate valid acts after multiple attempts.")
````

````python
from IPython.display import Markdown, display
import ipywidgets as widgets

import json

from model_text import display_messages
from llama_index.core.llms import ChatMessage
from model_text import llm_writer, stream_llm_response
from utils import deindent, ProgressTracker
from pydantic import ValidationError

# Maximum number of iterations
MAX_ITERATIONS = 20

def generate_chapters_and_scenes_per_act(story: Story, act: Act) -> Act:
    """
    Generates a list of chapters and scenes for an act.
    """

    system_message = deindent(f"""
        You are a creative story planner. Your task is to generate a structured list of chapters and scenes for the act. Return only a valid json object with the acts key.
        
        Each act should have:
        - chapters (list of chapters within the act)

        Each chapter should have:
        - chapter_id (string)
        - title
        - description
        - scenes (list)

        Each scene should include:
        - scene_id (string)
        - title
        - description
        - time_of_day
        - location
        - lighting
        - mood
        - setting
        - key actions
        - characters_involved_nicknames (list of character nicknames, nickname is unique to each character from the story characters list)
                              
        Consider:
        - The average chapter length ({story.avg_chapter_length}).
        - The average chapter count per act is: ({story.avg_chapters_per_act}).
        - Whether the act maintains a cohesive narrative and sufficient pacing.

        Return only a valid JSON object with the updated chapters and scenes for the act. Do not include any additional commentary or explanations.
    """)

    # Generate a markdown list of character nicknames to names
    character_list_md = "\n\n### Character Nicknames to Names:\n\n"
    for character in story.characters:
        character_list_md += f"- **{character.nickname}**: {character.name}\n"
        
    system_message += character_list_md

    messages = [ChatMessage(role="system", content=system_message)]

    for example in [few_shot_examples[0], few_shot_examples[-1]]:
        messages.append(ChatMessage(role="assistant", content="What story do you want to generate chapters and scenes for?"))
        messages.append(ChatMessage(role="user", content=example.model_dump_json(exclude={"acts": True})))
        messages.append(ChatMessage(role="assistant", content="Which act do you want to generate chapters and scenes for?"))
        messages.append(ChatMessage(role="user", content=example.acts[0].model_dump_json(exclude={"chapters": True})))
        messages.append(ChatMessage(role="assistant", content=example.acts[0].model_dump_json()))

    display_messages(messages)
    
    progress = ProgressTracker(max_items=600, description="Scene") # Approximate words for the response

    for attempt in range(MAX_RETRIES):
        progress.description = f"Attempt {attempt + 1} for Act: {act.title}"

        messages.append(ChatMessage(role="assistant", content="What story do you want to generate chapters and scenes for?"))
        messages.append(ChatMessage(role="user", content=story.model_dump_json(exclude={"acts": True})))
        messages.append(ChatMessage(role="assistant", content="Which act do you want to generate chapters and scenes for?"))
        messages.append(ChatMessage(role="user", content=act.model_dump_json()))

        response = llm_writer.stream_chat(messages, max_new_tokens=100000)
        response_text = stream_llm_response(response, progress).strip("```json").strip("```").strip()

        # Attempt to parse and return the updated act with chapters and scenes
        try:
            updated_act = Act.model_validate_json(response_text)
            progress.finish()
            return updated_act
        except (json.JSONDecodeError, ValidationError) as e:
            # Tell the llm what they output was wrong and why
            messages.append(ChatMessage(role="assistant", content=response_text))
            error_message = f"You failed to return a correct Story JSON object. Try again. This was the error: {str(e)}"
            messages.append(ChatMessage(role="system", content=error_message))
            messages.append(ChatMessage(role="system", content=system_message))
            # progress.reset()
            # progress.value = 0
            print()
            display(Markdown(f"Bad JSON:\n\n```\n{response_text}\n```"))
            print(str(e))
            if attempt == MAX_RETRIES - 1:
                raise ValueError(error_message)


def generate_scenes(story: Story) -> Story:
    """
    Main function that loops through each act to ensure it has sufficient chapters and scenes.
    """
    overall_progress = ProgressTracker(max_items=len(story.acts), description="Overall")
    failed_attempts = []
    
    for act_idx, act in enumerate(story.acts, start=1):
        # Retry if it fails to generate the act
        for iteration in range(MAX_ITERATIONS):
            display(Markdown(f"## Iteration {iteration + 1}: Act {act_idx} - '{act.title}'"))
            act = generate_chapters_and_scenes_per_act(story, act)
            try:
                story.acts[act_idx - 1] = act  # Update the act in the story
                display(Markdown(f":white_check_mark: Act {act_idx} completed successfully with {len(act.chapters)} chapters."))
                break
            except ValueError as e:
                display(Markdown(f"**Error during generation for Act {act_idx}:** {str(e)}"))
                
                failed_attempts.append(f"Act {act_idx} - '{act.title}'")
                if iteration == MAX_ITERATIONS - 1:
                    display(Markdown(f":warning: Reached maximum iterations ({MAX_ITERATIONS}) for Act {act_idx} without successful completion."))
                    raise
        overall_progress.update()

    overall_progress.finish()

    return story
````
